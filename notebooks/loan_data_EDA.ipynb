```python
for col in missing_cols:
    df[col + "_missing"] = df[col].isnull().astype(int)
```

```python
from scipy.stats import chi2_contingency

results = []
crosstab_results = {}

for col in missing_cols:
    indicator = col + "_missing"

    # Crosstab (percentage)
    ctab = pd.crosstab(df[indicator], df["Default"], normalize="index") * 100
    crosstab_results[col] = ctab

    # Chi-square
    table = pd.crosstab(df[indicator], df["Default"])
    chi2, p, dof, expected = chi2_contingency(table)

    results.append({
        "Feature": col,
        "p_value": p
    })

    print(f"\n==== {col} ====")
    print(ctab)

chi_df = pd.DataFrame(results).sort_values(by="p_value")
```

```python
chi_df["Significant"] = chi_df["p_value"] < 0.05

important_missing_features = chi_df[chi_df["Significant"]]["Feature"].tolist()
important_missing_features
```

```python
for col in missing_cols:
    if col not in important_missing_features:
        df.drop(columns=[col + "_missing"], inplace=True)
```

```python
# BINARY
# Apply to both train and test
for df_ in [X_train, X_test]:
    df_["Loan_Contract_Type"] = df_["Loan_Contract_Type"].map({"CL":0, "RL":1, "Unknown":0})
    df_["Client_Permanent_Match_Tag"] = df_["Client_Permanent_Match_Tag"].map({"Yes":1, "No":0})
    df_["Client_Contact_Work_Tag"] = df_["Client_Contact_Work_Tag"].map({"Yes":1, "No":0})

# ONE-HOT
one_hot_cols = [
    "Accompany_Client",
    "Client_Income_Type",
    "Client_Marital_Status",
    "Client_Gender",
    "Client_Housing_Type",
    "Client_Education"
]

X_train = pd.get_dummies(X_train, columns=one_hot_cols, drop_first=True)
X_test = pd.get_dummies(X_test, columns=one_hot_cols, drop_first=True)

# FREQUENCY
for col in ["Client_Occupation", "Type_Organization"]:
    freq = X_train[col].value_counts(normalize=True)

    X_train[col] = X_train[col].map(freq)
    X_test[col] = X_test[col].map(freq)
```

```python
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)
```

```python
from imblearn.over_sampling import SMOTE, RandomOverSampler
from sklearn.metrics import roc_auc_score, recall_score, precision_score, f1_score

samplers = {
    "No Sampling": None,
    "SMOTE": SMOTE(random_state=42),
    "Oversampling": RandomOverSampler(random_state=42)
}

results = []

for name, sampler in samplers.items():

    X_tr, y_tr = X_train.copy(), y_train.copy()

    if sampler:
        X_tr, y_tr = sampler.fit_resample(X_tr, y_tr)

    model.fit(X_tr, y_tr)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:,1]

    results.append({
        "Method": name,
        "ROC_AUC": roc_auc_score(y_test, y_prob),
        "Recall": recall_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred)
    })

import pandas as pd
pd.DataFrame(results).sort_values(by="ROC_AUC", ascending=False)
```

```python
group_map = {
    "Accompany_Client": [col for col in X_train.columns if col.startswith("Accompany_Client_")],
    "Client_Income_Type": [col for col in X_train.columns if col.startswith("Client_Income_Type_")],
    "Client_Marital_Status": [col for col in X_train.columns if col.startswith("Client_Marital_Status_")],
    "Client_Gender": [col for col in X_train.columns if col.startswith("Client_Gender_")],
    "Client_Housing_Type": [col for col in X_train.columns if col.startswith("Client_Housing_Type_")],
    "Client_Education": [col for col in X_train.columns if col.startswith("Client_Education_")]
}
```

```python
group_importance = {}

# 1. Aggregate grouped features
for group, cols in group_map.items():
    existing_cols = [c for c in cols if c in importance.index]
    group_importance[group] = importance[existing_cols].sum()

# 2. Add non-grouped features
for col in importance.index:
    if not any(col.startswith(group + "_") for group in group_map):
        group_importance[col] = importance[col]
```

```python
group_imp_df = pd.Series(group_importance).sort_values(ascending=False)

print("\nAggregated Feature Importance:")
print(group_imp_df.head(15))
```

```python
import matplotlib.pyplot as plt

group_imp_df.head(15).plot(kind="barh", figsize=(8,5))
plt.gca().invert_yaxis()
plt.title("Top 15 Aggregated Feature Importance")
plt.xlabel("Importance")
plt.show()
```
